```python
import numpy as np

def create_polynomial_features(x, degree):
  m = len(x)
  X_poly = np.zeros((m,degree))

  for d in range(1, degree+1):
    X_poly[:,d-1] = np.power(x,d)

  return X_poly


x = np.array([2,4,6])
degree = 3
X_poly = create_polynomial_features(x, degree)
print(X_poly)
```

```python
def create_polynomial_features(x, degree):
    powers = np.arange(1, degree + 1)
    return x.reshape(-1, 1) ** powers

x = np.array([2,4,6])
degree = 3
X_poly = create_polynomial_features(x, degree)
print(X_poly)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 50)

X = np.column_stack((x, x**2, x**3))

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(x, X[:, 0])
plt.title('x')

plt.subplot(1, 3, 2)
plt.plot(x, X[:, 1])
plt.title('x²')

plt.subplot(1, 3, 3)
plt.plot(x, X[:, 2])
plt.title('x³')

plt.show()
```
