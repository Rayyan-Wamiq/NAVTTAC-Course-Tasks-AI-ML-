```python
import numpy as np

def one_hot_encode(y, num_classes):
    m = len(y)
    encoded = np.zeros((m, num_classes))
    encoded[np.arange(m), y] = 1
    return encoded

y = np.array([0,1,2,0,1,2,0,2,1,2,1,0])
print(one_hot_encode(y, num_classes=3))
```
