```python
import numpy as np

def gradient_descent(X, y, w, b, alpha=0.1, iterations=500):
    for _ in range(iterations):
        z = X @ w + b
        p = 1 / (1 + np.exp(-z))

        dw = (X.T @ (p - y)) / len(y)
        db = np.mean(p - y)

        w -= alpha * dw
        b -= alpha * db

    return w, b, None

num_classes = 3
models = []

X = np.array([
    [1, 2, 4, 3],
    [2, 3, 4, 5],
    [5, 2, 1, 4]
])

y = np.array([0, 1, 2])

for c in range(num_classes):
    y_binary = (y == c).astype(int)
    w = np.zeros(X.shape[1])
    b = 0

    w, b, _ = gradient_descent(X, y_binary, w, b, 0.1, 500)
    models.append((w, b))

print(models)
```

```python
import numpy as np

def hypothesis(x, w, b):
    return 1 / (1 + np.exp(-(x @ w + b)))


def predict_ovr(x, models):
    probabilities = np.array([
        hypothesis(x, w, b) for w, b in models
    ])
    return np.argmax(probabilities, axis=0)


x = np.array([
    [1, 2, 4, 3],
    [2, 3, 4, 5],
    [5, 2, 1, 4]
])

prediction = predict_ovr(x, models)

print("Prediction:", prediction)
```
