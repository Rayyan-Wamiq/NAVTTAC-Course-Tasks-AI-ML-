```python
import pandas as pd
df = pd.DataFrame({
    "region": ["Asia", "Europe", "Mena","North America"],
    "sales": [50000, 55000, 70000, 72000]
})
result = df.groupby("region")["sales"].agg(["mean", "min", "max", "count"])
print(result)

```

```python
import pandas as pd
df = pd.DataFrame({
    "region": ["Asia", "Europe", "Mena","North America"],
    "sales": [50000, 55000, 70000, 72000],
    "yearlysales": [100000, 500000, 100000, 450000]
})

result = df.groupby("region")["sales"].sum()
print(result)
```

```python
ages = pd.Series([5, 17, 22, 45, 67, 34, 12, 80])

bins = [0, 12, 19, 64, 100]
labels = ["Child", "Teen", "Adult", "Senior"]

categories = pd.cut(ages, bins=bins, labels=labels)
print(categories)
print(categories.value_counts())
```

```python
scores = pd.Series([55, 60, 62, 70, 75, 80, 85, 90, 95, 100])

quartile_bins = pd.qcut(scores, q=4, labels=["Q1", "Q2", "Q3", "Q4"])
print(quartile_bins)
print(quartile_bins.value_counts())
```
