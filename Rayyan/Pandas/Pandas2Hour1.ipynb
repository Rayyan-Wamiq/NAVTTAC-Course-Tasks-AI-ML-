```python
import pandas as pd

jan = pd.DataFrame({"product": ["A", "B"], "sales": [100, 150]})
feb = pd.DataFrame({"product": ["A", "B"], "sales": [120, 130]})

combined = pd.concat([jan, feb], ignore_index = True)
print(combined)
```

```python
from numpy import nan
Student = pd.DataFrame({
    "student_id": [1, 2, 3, 4],
    "name": ["Alice", "Bob", "Charlie", "David"]
})
Grades = pd.DataFrame({
    "student_id": [1, 2, 3, 3],
    "grade": [85, 92, 78, nan]
})

result = pd.merge(Student, Grades, on="student_id")
print(result)
```

```python
from numpy import nan
Student = pd.DataFrame({
    "student_id": [1, 2, 3, 4],
    "name": ["Alice", "Bob", "Charlie", "David"]
})
Grades = pd.DataFrame({
    "student_id": [1, 2, 3, 3],
    "grade": [85, 92, 78, nan]
})

result = pd.merge(Student,Grades, on="student_id", how="left")
print(result)


result = pd.merge(Student,Grades, on="student_id", how="right")
print(result)


result = pd.merge(Student,Grades, on="student_id", how="outer")
print(result)
```

```python
Products = pd.DataFrame({
    "product_id": [1, 2, 3],
    "product_name": ["Shampoo", "Soap", "Surf"],
    "price": [750, 250, 400]
})

Orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "product_id": [2, 1, 4],
    "quantity": [2, 1, 3]
})

result = pd.merge(Products, Orders, on="product_id")
print(result)

result = pd.merge(Products, Orders, on="product_id", how="right")
print(result)
```
