```python
def hypothesis(x, w, b):
    return w * x + b


def compute_cost(x, y, w, b):
    m = len(x)
    total_squared_error = 0
    squared_errors = []

    for i in range(m):
        prediction = hypothesis(x[i], w, b)
        error = prediction - y[i]
        squared_error = error ** 2

        squared_errors.append(squared_error)
        total_squared_error += squared_error

    cost = total_squared_error / m

    return cost, squared_errors


x_train = [1, 2, 3]
y_train = [4, 5, 6]

cost, errors = compute_cost(x_train, y_train, w=5, b=45)

print("Cost:", cost)
print("Squared Errors:", errors)
```
