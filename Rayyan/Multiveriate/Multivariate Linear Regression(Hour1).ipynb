```python
import numpy as np

w = np.array([100, 5000, 8000, -1500])   # weight per feature
x = np.array([1800, 4, 2, 15])         # one house's feature values
b = 20000

# h(x) = w . x + b  = w_1*x_1 + w_2*x_2 + ... + w_n*x_n + b  <-- This line is mathematical notation and not valid Python code.
prediction = np.dot(w, x) + b
print(prediction)
```
