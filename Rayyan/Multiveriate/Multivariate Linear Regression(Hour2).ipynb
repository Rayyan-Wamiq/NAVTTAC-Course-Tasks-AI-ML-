```python
import numpy as np
def hypothesis(X, w, b):
    return np.dot(X, w) + b

w = np.array([150, 8000, -2000])
b = 30000

predictions = hypothesis(X, w, b)
print(predictions)


def compute_cost(X, y, w, b):
    m = X.shape[0]
    predictions = hypothesis(X, w, b)
    errors = predictions - y
    cost = np.sum(errors ** 2) / m
    return cost

print(compute_cost(X, y, w, b))

def compute_gradients(X, y, w, b):
    m = X.shape[0]
    predictions = hypothesis(X, w, b)
    errors = predictions - y

    dw = np.dot(X.T, errors) / m    # X.T -- gives ALL n gradients at once
    db = np.sum(errors) / m

    return dw, db

```
