```python
'''17. Rule-based NLP systems struggled to scale because they depended on humans writing rules for every possible situation. Natural language is full of ambiguity and exceptions. For example, “I saw her duck” can mean seeing her lower her head or seeing a duck that belongs to her. A system would need many rules to determine which meaning is correct based on context. As language became more complex, the number of rules became extremely large and difficult to maintain. Small changes or new expressions could also cause errors.

18. Modern neural NLP models still use tokenization because they need to break language into manageable pieces before processing it. Instead of treating an entire sentence as one object, tokenization divides it into words, subwords, or sometimes individual characters. This gives the model smaller units that can be converted into numerical representations and analyzed. Even though neural models can learn context and meaning much better than rule-based systems, they still need some structured representation of the input before they can process it.'''
```
