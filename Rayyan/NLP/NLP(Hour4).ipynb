```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab", force=True)
from nltk.tokenize import word_tokenize, sent_tokenize

text = "NLP is fascinating. It powers chatbots and translators!"

print(sent_tokenize(text))

print(word_tokenize(text))
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp("NLP is fascinating. It powers chatbots and translators!")

for token in doc:
    print(token.text, token.pos_, token.is_stop)
```

```python
from nltk.corpus import stopwords
nltk.download("stopwords")

stop_words = set(stopwords.words("english"))
words = word_tokenize("This is a simple example of removing stopwords")

filtered = [w for w in words if w.lower() not in stop_words]
print(filtered)
```

```python
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Make sure these are downloaded first:
# import nltk
# nltk.download("stopwords")

stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()

def clean_text(text):
    # 36. Lowercasing
    text = text.lower()

    # 37. Punctuation removal
    text = re.sub(r"[^\w\s]", "", text)

    # 38. Tokenization
    tokens = text.split()

    # 39. Stopword removal
    tokens = [word for word in tokens if word not in stop_words]

    # 40. Stemming
    tokens = [stemmer.stem(word) for word in tokens]

    return tokens


# 41. Test with a messy sample paragraph
sample = """WOW! The students are learning NLP, and they're REALLY
enjoying it. The lessons aren't difficult; they're interesting and useful!"""

cleaned = clean_text(sample)
print(cleaned)
```
